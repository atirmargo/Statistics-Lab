1. X1, ..., Xn ∼ Bern(θ), θ ∼ R[0, 1]. Посчитать апостериорную плотность и построить ее график.

а) для выборок размера n = 5, 10, 20, 50, 100 для θ = 1/2, θ = 1/3;

б) вместо генерации выборки положите sum(Xi) равной 9n/10 или 99n/100.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

def find_posterior_density(n_values, theta_values): #Пункт а
    theta_plot = np.linspace(0, 1, 200)
    for n in n_values:
        for theta in theta_values:
            sample = np.random.binomial(1, theta, n)
            k = np.sum(sample)
            likelihood = (theta**k) * ((1 - theta)**(n-k))
            alpha = k + 1
            beta = n - k + 1
            posterior = stats.beta.pdf(theta_plot, alpha, beta)
            print(f"Апостериорная плотность Beta{int(alpha), int(beta)}, размер выборки {n}, theta_true = {theta}")            
            plt.figure(figsize=(5, 3))
            plt.plot(theta_plot, posterior)   
            plt.xlabel('theta') 
            plt.ylabel('Апостериорная плотность')
            plt.axvline(x = theta, color = 'red')
            plt.show()

def find_posterior_density_2(n_values, theta_values): #Пункт б
    theta_plot = np.linspace(0, 1, 200)
    for n in n_values:
        for theta in theta_values:
            k = 99 * n / 100
            likelihood = (theta**k) * ((1 - theta)**(n-k))
            alpha = k + 1
            beta = n - k + 1
            posterior = stats.beta.pdf(theta_plot, alpha, beta)
            print(f"Апостериорная плотность Beta{int(alpha), int(beta)}, размер выборки {n}, theta_true = {theta}")            
            plt.figure(figsize=(5, 3))
            plt.plot(theta_plot, posterior)   
            plt.xlabel('theta') 
            plt.ylabel('Апостериорная плотность')
            plt.axvline(x = theta, color = 'red')
            plt.show()
            
n_values = [5, 10, 20, 50, 100]
theta_values = [1/2, 1/3]
find_posterior_density(n_values, theta_values)  
print('------------------------------------------------------')
find_posterior_density_2(n_values, theta_values)     
```

2. Пусть X1, ..., Xn ∼ N (θ, 1), θ ∼ N (μ, σ2). Апостериорную плотность можно не считать, а взять из
таблицы. Посмотрим, как влияют параметры априорного распределения на итоговый результат.
Для этого будем генерировать выборки для какого-то одного фиксированного θ (возьмите любое
число из [0, 1], а также посмотрите на какое-нибудь θ, близкое к 0 или 1).

(a) Сравнить (визуально на графике) апостериорные плотности для нескольких разных значений (μ, σ2).

(b) Построить пример, когда при n = 10000 оценка θ достаточно сильно отличается от настоящего значения.

(c) Сравнить байесовские оценки для квадратичного риска при разных (μ, σ2).

```python
#Апострериорная плотность для X1, ..., Xn ∼ N (θ, 1), θ ∼ N (μ, σ2) это N(μ_n, σ_n²)
#         μ/σ² + n·x̄
# μ_n = -------------
#         1/σ² + n
#           1
# σ_n² = -------
#         1/σ² + n

def function(theta_values, values, n, x):
    colors = ['orange', 'green', 'red']
    for theta in theta_values:
        sample = np.random.normal(theta, 1, n)
        errors = []
        for idx, (mean, variance) in enumerate(values):
            mean_n = (mean/variance + n * np.mean(sample)) / (1/variance + n) #Это и есть байесовская оценка
            variance_n = 1 / (1/variance + n)
            y = stats.norm.pdf(x, mean_n, variance_n)
            errors.append((mean_n - theta)**2)
            print(f"(μ, σ²) = {(mean, variance)}, μ_n = {mean_n}, σ_n² = {variance_n}, theta = {theta}")
            plt.plot(x, y, color = colors[idx])
        plt.axvline(theta, color = 'black', linestyle='--', label = "true_theta")
        plt.axvline(np.mean(sample), color = 'blue', linestyle='--', label = "mean(sample)")
        plt.axvline(mean_n, color = 'brown', label = "bayesian_est")
        plt.xlabel('theta')
        plt.ylabel('Апостериорная плотность')
        plt.legend()
        plt.show() 
        print(f"Сравнение байесовских оценок: \n ошибка для первой {errors[0]} \n ошибка для второй {errors[1]} \n ошибка для третьей {errors[2]}")
def find_bad_example(theta, n):
    N_simulations = 1000
    estimates = []
    for i in range(N_simulations):
        sample = np.random.normal(theta, 1, n)
        theta_est = np.mean(sample)
        estimates.append(theta_est)
        if abs(theta - theta_est) > 0.03: #3 сигмы
            print(f"Оценка theta {theta_est} отличается от theta > чем на 0.03 ")
            plt.plot(sample[:200], 'o', markersize=3, alpha=0.5)
            plt.axhline(theta, color='red', label=f'true_theta = {theta}')
            plt.axhline(theta_est, color='blue', linestyle='--', label=f'Оценка theta = {theta_est}')
            plt.legend()
            plt.show()
            break
  
theta_values = [0.1, 0.5, 0.9]
values = [(0,1), (0.5, 0.25), (0, 10)]
x = np.linspace(-2, 2, 200)
n = 5
function(theta_values, values, n, x)
find_bad_example(1/2, 10000)
```

3. X1, ..., Xn ∼ N (θ, 1). Построить (на листочке) байесовский критерий для проверки H0 : θ = 0
против H1 : θ = 1, если априорная вероятность P(θ = 0) = p. Построить графики зависимости
ошибок 1-го и 2-го рода от p.
![байесовская оценка](bayes_estimation.png)

```python
def plots_of_type_1_and_type_2_errors(n, p_values, L1 = 1, L2 = 1):
    type_1_errors = []
    type_2_errors = []
    #sample = np.random.nolmal(theta, 1, n) -> Критерий имеет вид D = {np.mean(sample) > C}
    for p in p_values:
        c = (p * L1) / ((1 - p) * L2)
        C = np.log(c) / n + 1/2
        type_1_error = 1 - stats.norm.cdf(np.sqrt(n) * C)
        type_2_error = stats.norm.cdf(np.sqrt(n) * (1 - C))
        type_1_errors.append(type_1_error)
        type_2_errors.append(type_2_error)
         
    plt.plot(p_values, type_1_errors, color = 'red', label = "Вероятность ошибки 1го рода")
    plt.plot(p_values, type_2_errors, color = 'blue', label = "Вероятность ошибки 2го рода")
    plt.legend()
    plt.show()

p_values = np.linspace(0.01, 0.99, 100)
n = 100
plots_of_type_1_and_type_2_errors(n, p_values, L1 = 1, L2 = 1)
```
